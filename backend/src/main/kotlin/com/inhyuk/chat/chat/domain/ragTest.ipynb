```kotlin
import dev.langchain4j.data.document.loader.FileSystemDocumentLoader
import dev.langchain4j.data.document.parser.apache.pdfbox.ApachePdfBoxDocumentParser
import dev.langchain4j.data.document.parser.apache.poi.ApachePoiDocumentParser
import dev.langchain4j.data.document.parser.apache.tika.ApacheTikaDocumentParser
import dev.langchain4j.model.googleai.GoogleAiGeminiChatModel

import dev.langchain4j.service.AiServices
import dev.langchain4j.service.UserMessage
import dev.langchain4j.service.V
import java.nio.file.Paths

val pdfParser: ApachePdfBoxDocumentParser = ApachePdfBoxDocumentParser()
val tikaParser: ApacheTikaDocumentParser = ApacheTikaDocumentParser()
val poiParser = ApachePoiDocumentParser()
val document = FileSystemDocumentLoader.loadDocument("D:\\private\\ppr\\chat-ih\\backend\\upload\\요구사항정의목록.xlsx", poiParser)

val model = GoogleAiGeminiChatModel.builder().apiKey("").modelName("gemini-3-flash-preview").build()

interface AiAssistant {
    @UserMessage("{{userMessage}}{{attachments}}")
    fun chat(@V("userMessage") message: String?, @V("attachments") attachments: String?): String

}


val assistant = AiServices.create(AiAssistant::class.java,model)

val response = assistant.chat("안녕하세요. 첨부한 파일의 내용을 보고 목표사이클에 대해서 알려주세요.","""
<attachments>
    <file name="요구사항정의목록.xlsx" type="xlsx">
        ${document.text()}
    </file>
</attachments>
""")

println(response)
```
